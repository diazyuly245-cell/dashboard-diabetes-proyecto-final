```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score, r2_score, mean_squared_error

# --- DATOS ---
data = load_diabetes(scaled=False)
df = pd.DataFrame(data.data, columns=data.feature_names)
df['progresion'] = data.target

# Diccionario para traducir los nombres de columna a etiquetas en español
nombres_espanol = {
    'age': 'Edad',
    'sex': 'Sexo',
    'bmi': 'IMC (Índice de Masa Corporal)',
    'bp': 'Presión arterial',
    's1': 'Colesterol total (s1)',
    's2': 'LDL - colesterol malo (s2)',
    's3': 'HDL - colesterol bueno (s3)',
    's4': 'Colesterol total/HDL (s4)',
    's5': 'Triglicéridos (s5)',
    's6': 'Glucosa (s6)',
}

# --- MODELO DE REGRESION LINEAL ---
X_reg = df[['bmi']].values
y_reg = df['progresion'].values
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.30, random_state=42)
modelo_lineal = LinearRegression()
modelo_lineal.fit(X_train_reg, y_train_reg)
y_pred_reg = modelo_lineal.predict(X_test_reg)
r2_lineal = r2_score(y_test_reg, y_pred_reg)
mse_lineal = mean_squared_error(y_test_reg, y_pred_reg)

# --- MODELO DE REGRESION LOGISTICA ---
mediana = df['progresion'].median()
df['progresion_alta'] = (df['progresion'] > mediana).astype(int)
X_log = df[['bmi']].values
y_log = df['progresion_alta'].values
X_train_log, X_test_log, y_train_log, y_test_log = train_test_split(X_log, y_log, test_size=0.30, random_state=42, stratify=y_log)
modelo_log = LogisticRegression()
modelo_log.fit(X_train_log, y_train_log)
y_proba_log = modelo_log.predict_proba(X_test_log)[:, 1]

# --- COLORES ---
COLOR_PRIMARIO = '#0f3d6e'
COLOR_ACENTO = '#e07b1f'
COLOR_FONDO = '#f7f7f7'
estilo_tab = {'padding': '10px', 'fontWeight': 'bold'}
estilo_tab_seleccionada = {'padding': '10px', 'fontWeight': 'bold', 'backgroundColor': COLOR_PRIMARIO, 'color': 'white'}

# --- APP ---
app = dash.Dash(__name__)
app.config.suppress_callback_exceptions = True

app.layout = html.Div([
    html.H1("Dashboard: IMC y Progresión de la Diabetes", style={'textAlign': 'center', 'color': COLOR_PRIMARIO, 'padding': '20px'}),
    dcc.Tabs(id="tabs", value='tab-contexto', children=[
        dcc.Tab(label='Contexto', value='tab-contexto', style=estilo_tab, selected_style=estilo_tab_seleccionada),
        dcc.Tab(label='Exploración', value='tab-exploracion', style=estilo_tab, selected_style=estilo_tab_seleccionada),
        dcc.Tab(label='Regresión', value='tab-regresion', style=estilo_tab, selected_style=estilo_tab_seleccionada),
        dcc.Tab(label='Clasificación', value='tab-clasificacion', style=estilo_tab, selected_style=estilo_tab_seleccionada),
        dcc.Tab(label='Conclusiones', value='tab-conclusiones', style=estilo_tab, selected_style=estilo_tab_seleccionada),
    ]),
    html.Div(id='contenido-tab', style={'padding': '25px'})
], style={'fontFamily': 'Arial, sans-serif', 'backgroundColor': COLOR_FONDO, 'minHeight': '100vh'})
```

```python
@app.callback(Output('contenido-tab', 'children'), Input('tabs', 'value'))
def render_contenido(pestaña_activa):
    if pestaña_activa == 'tab-contexto':
        return html.Div([
            html.H2("Contexto del problema"),
            html.H3("Campo de estudio"),
            html.P("Salud pública / epidemiología clínica — análisis de factores de riesgo en enfermedades crónicas no transmisibles, con foco en la diabetes."),
            html.H3("Pregunta de investigación"),
            html.P("En pacientes diagnosticados con diabetes, ¿en qué medida el Índice de Masa Corporal (IMC) explica la variabilidad en la progresión de la enfermedad un año después del diagnóstico?"),
            html.H3("Por qué es relevante"),
            html.Ul([
                html.Li("El IMC es un factor de riesgo modificable, a diferencia de la edad o la genética."),
                html.Li("Medir el IMC es simple y económico, ideal para contextos con recursos limitados."),
                html.Li("Entender los límites de esta relación evita sobreestimar la capacidad de un solo indicador."),
            ]),
            html.H3("Sobre el dataset"),
            html.P("Dataset de diabetes de scikit-learn: 442 pacientes, 10 variables predictoras más la progresión de la enfermedad."),
            html.P("Fuente: Efron, Hastie, Johnstone & Tibshirani (2004). Least Angle Regression. Annals of Statistics.", style={'fontStyle': 'italic', 'fontSize': '14px'}),
        ])
    elif pestaña_activa == 'tab-exploracion':
        return html.Div([
            html.H2("Exploración de variables"),
            html.P("Elige una variable para comparar contra la progresión de la enfermedad:"),
                        dcc.Dropdown(
                id='variable-dropdown',
                options=[{'label': nombres_espanol[col], 'value': col} for col in data.feature_names],
                value='bmi'
            ),
            
            dcc.Graph(id='grafico-exploracion')
        ])
    elif pestaña_activa == 'tab-regresion':
        fig_reg = px.scatter(df, x='bmi', y='progresion', trendline='ols', template='plotly_white',
                              color_discrete_sequence=[COLOR_ACENTO], title='IMC vs. Progresión, con línea de regresión')
        return html.Div([
            html.H2("Regresión lineal: IMC → Progresión"),
            dcc.Graph(figure=fig_reg),
            html.P(f"Coeficiente: {modelo_lineal.coef_[0]:.2f}"),
            html.P(f"R² (conjunto de prueba): {r2_lineal:.4f}"),
            html.P(f"MSE (conjunto de prueba): {mse_lineal:.2f}"),
        ])
    elif pestaña_activa == 'tab-clasificacion':
        return html.Div([
            html.H2("Regresión logística: clasificación de progresión alta/baja"),
            html.P("Mueve el slider para cambiar el umbral de decisión:"),
            dcc.Slider(id='umbral-slider', min=0.05, max=0.95, step=0.05, value=0.5,
                       marks={i/10: str(i/10) for i in range(1, 10)}),
            html.Div(id='resultado-clasificacion')
        ])
    elif pestaña_activa == 'tab-conclusiones':
        return html.Div([
            html.H2("Conclusiones principales"),
            html.Ul([
                html.Li("Existe una relación estadísticamente significativa entre el IMC y la progresión (p-value = 3.47×10⁻⁴²)."),
                html.Li("El IMC explica el 28% de la variabilidad de la progresión (R²=0.28)."),
                html.Li("Usando solo el IMC, se clasifica correctamente a 7 de cada 10 pacientes (accuracy=70.7%)."),
                html.Li("El IMC es un factor relevante pero parcial de una enfermedad multifactorial.")
            ])
        ])


@app.callback(Output('grafico-exploracion', 'figure'), Input('variable-dropdown', 'value'))
def actualizar_grafico(variable_elegida):
    fig = px.scatter(df, x=variable_elegida, y='progresion', trendline='ols', template='plotly_white',
                      color_discrete_sequence=[COLOR_ACENTO], title=f'{variable_elegida} vs. Progresión de la enfermedad')
    return fig


@app.callback(Output('resultado-clasificacion', 'children'), Input('umbral-slider', 'value'))
def actualizar_clasificacion(umbral):
    y_pred_umbral = (y_proba_log >= umbral).astype(int)
    acc = accuracy_score(y_test_log, y_pred_umbral)
    cm = confusion_matrix(y_test_log, y_pred_umbral)
    tn, fp, fn, tp = cm.ravel()
    fig_cm = px.imshow(cm, text_auto=True, color_continuous_scale='Blues',
                        labels=dict(x="Predicho", y="Real", color="Cantidad"),
                        x=['Baja (0)', 'Alta (1)'], y=['Baja (0)', 'Alta (1)'],
                        title=f"Matriz de confusión (umbral = {umbral})")
    return html.Div([
        dcc.Graph(figure=fig_cm),
        html.P(f"Accuracy: {acc:.4f} | Falsos positivos: {fp} | Falsos negativos: {fn}")
    ])
```

```python
app.run(debug=True, port=8060)
```
